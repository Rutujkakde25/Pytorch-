```
import torch
import torch.nn as nn
```

```
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.network=nn.Sequential(
      nn.Linear(num_features,3),
      nn.ReLU(),
      nn.Linear(3,1),
      nn.Sigmoid()

      )
  def forward(self,x):
         return self.network(x)


```

```
x=torch.randn(10,5)
model = Model(x.shape[1])
model(x)
```
