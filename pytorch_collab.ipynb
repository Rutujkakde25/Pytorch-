```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print("y =", y)
print("dy/dx =", x.grad)
```

```python
import torch

# ============================================
# STEP 1: Create Training Data
# ============================================

# Input (x)
x = torch.tensor([1.0, 2.0, 3.0, 4.0])

# Correct Output (y)
y_true = torch.tensor([2.0, 4.0, 6.0, 8.0])

# ============================================
# STEP 2: Initialize Weight
# ============================================

# Start with a random weight
w = torch.tensor(0.0, requires_grad=True)

# ============================================
# STEP 3: Hyperparameters
# ============================================

learning_rate = 0.01
epochs = 20

# ============================================
# STEP 4: Training Loop
# ============================================

for epoch in range(epochs):

    # -----------------------------
    # Forward Pass
    # -----------------------------
    y_pred = w * x

    # -----------------------------
    # Calculate Loss (Mean Squared Error)
    # -----------------------------
    loss = ((y_pred - y_true) ** 2).mean()

    # -----------------------------
    # Backward Pass
    # Compute Gradient
    # -----------------------------
    loss.backward()

    # -----------------------------
    # Update Weight
    # -----------------------------
    with torch.no_grad():
        w -= learning_rate * w.grad

    # -----------------------------
    # Print Progress
    # -----------------------------
    print(
        f"Epoch {epoch+1:2d} | "
        f"Weight = {w.item():.4f} | "
        f"Loss = {loss.item():.4f}"
    )

    # -----------------------------
    # Clear Gradient
    # -----------------------------
    w.grad.zero_()

# ============================================
# Final Weight
# ============================================

print("\nTraining Finished")
print("Final Weight =", w.item())
```

1. loss.backward()
Definition

loss.backward() starts the backward pass. It tells PyTorch's Autograd engine to compute the gradients (derivatives) of the loss with respect to all tensors that have requires_grad=True.

What happens internally?
PyTorch traverses the computation graph from the loss back to the parameters.
It applies the Chain Rule to compute derivatives.
The computed gradients are stored in the .grad attribute of each parameter.

Example:

loss.backward()

print(w.grad)

Output:

tensor(-16.)

Here, w.grad contains:

d(Loss) / d(Weight)
2. w.grad
Definition

w.grad stores the gradient (derivative) of the loss with respect to the weight.

Mathematically:

w.grad = d(Loss) / dw
Purpose

The gradient tells us:

How should the weight change to reduce the loss?

Remember:

w = Weight (parameter)
w.grad = Slope of the loss function

Example:

Weight = 3

Gradient = -16

The weight is still 3. The gradient only tells us how to update it.

3. Why is the Gradient Needed?

A neural network learns by changing its weights.

The gradient answers the question:

"If I change this weight slightly, how will the loss change?"

Positive gradient → Decrease the weight
Negative gradient → Increase the weight
Zero gradient → Already near the minimum loss

Without gradients, the model cannot learn.

4. Learning Rate
learning_rate = 0.1
Definition

The learning rate controls how big each update step should be.

Formula:

New Weight = Old Weight − Learning Rate × Gradient

Example:

Old Weight = 3

Gradient = -16

Learning Rate = 0.1

New Weight

= 3 − (0.1 × -16)

= 4.6
Small Learning Rate

Advantages:

Stable training
Accurate learning

Disadvantages:

Slow training
Large Learning Rate

Advantages:

Faster learning

Disadvantages:

Can overshoot the minimum
Training may become unstable
5. Weight Update
w -= learning_rate * w.grad

Equivalent to:

w = w - learning_rate * w.grad
Purpose

This updates the weight using the gradient computed by Autograd.

This process is called Gradient Descent.

Without updating the weights, the model will never improve.

6. torch.no_grad()
with torch.no_grad():
Definition

Temporarily disables gradient tracking.

Why is it needed?

During the forward pass, PyTorch records every operation to build the computation graph.

Example:

y = w * x

This operation is recorded because gradients will be needed later.

However, when updating the weight:

w -= learning_rate * w.grad

we are not performing part of the neural network computation.

We are simply changing the value of the parameter.

If PyTorch tracked this update, it would create unnecessary computation graphs, consume more memory, and could lead to incorrect behavior.

Therefore, we use:

with torch.no_grad():

which means:

"Do not track any operations inside this block."

Benefits
Saves memory
Faster execution
Prevents unnecessary computation graphs
7. w.grad.zero_()
w.grad.zero_()
Definition

Resets the gradient stored in w.grad to zero.

Why is it needed?

PyTorch accumulates gradients by default.

Example:

First backward pass:

Gradient = -16

Second backward pass:

Gradient = -32

Third backward pass:

Gradient = -48

because gradients are added together.

Before computing the next gradient, we must clear the previous one.

w.grad.zero_()

Now:

Gradient = 0

The next backward pass computes a fresh gradient.

8. Why Does PyTorch Accumulate Gradients?

Gradient accumulation is useful when training large models.

Example:

Instead of processing:

1024 images

at once, we process:

Batch 1
↓

Compute Gradient

↓

Batch 2

↓

Add Gradient

↓

Batch 3

↓

Add Gradient

↓

Update Weight Once

This technique is called Gradient Accumulation.

For normal training, however, we clear the gradients after every update.
